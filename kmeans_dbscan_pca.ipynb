## 2. Scale the Features

Clustering algorithms can be affected when features have very different ranges.

So, we standardize the data before clustering.


```python
# Scale the features

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data has been scaled.")
```

## 3. Elbow Method for K-Means

K-Means needs us to choose the number of clusters, called `k`.

The elbow method helps us choose a reasonable value.

Look for the point where the line starts bending like an elbow.


```python
# Try different values of k

inertia = []

for k in range(1, 15):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    inertia.append(model.inertia_)

plt.plot(range(1, 15), inertia, marker='o')
```

### Choose k

For this dataset, `k = 3` is a simple and reasonable choice from the elbow graph.


```python
# Apply K-Means with k = 4

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans.fit(X_scaled)
kmeans_labels = kmeans.fit_predict(X_scaled)

print("K-Means cluster labels:")
print(kmeans_labels[:20])
```

```python
kmeans_labels
```

## 4. Apply DBSCAN

DBSCAN does not need us to choose the number of clusters.

Instead, it groups points that are close together.

Points labelled `-1` are considered **noise or outliers**.


```python
# Apply DBSCAN

dbscan = DBSCAN(eps=2.5, min_samples=5)

dbscan_labels = dbscan.fit_predict(X_scaled)

print("DBSCAN cluster labels:")
print(dbscan_labels[:20])

print("\nUnique DBSCAN labels:", set(dbscan_labels))
```

```python
X_scaled.shape
```

## 5. Reduce the Data to 2 Dimensions with PCA

Our dataset has many features.

PCA reduces it to only 2 dimensions so that we can visualize the clusters.


```python
# Reduce the scaled data to 2 dimensions

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("New PCA shape:", X_pca.shape)
```

```python
X_pca
```

## 6. Visualize K-Means Clusters


```python
plt.figure(figsize=(7, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=kmeans_labels,
    cmap="viridis"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clusters")
plt.show()
```

## 7. Visualize DBSCAN Clusters

DBSCAN may produce a different number of groups from K-Means.

Remember: label `-1` represents noise.


```python
plt.figure(figsize=(7, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=dbscan_labels,
    cmap="viridis"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("DBSCAN Clusters")
plt.show()
```

## 8. Simple Comparison

### K-Means
- We must choose the number of clusters.
- Here we selected `k = 3` using the elbow method.
- Every data point is assigned to a cluster.

### DBSCAN
- We do not choose the number of clusters directly.
- It finds groups based on dense areas.
- It can identify noise or outliers using label `-1`.

### Main Idea

Different clustering algorithms can produce different groupings on the same dataset.

K-Means works well when clusters are fairly clear and compact.

DBSCAN is useful when we want to find dense groups and possible outliers.
